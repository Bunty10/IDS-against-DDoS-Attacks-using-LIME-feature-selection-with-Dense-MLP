```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("LIME_ALL_feature_importance.csv", low_memory=False)
```

```python
df.columns.values.tolist()
```

```python
val = df["LIME Importance"]
```

```python
mean = np.mean(val)
```

```python
mean
```

```python
median = np.median(val)
```

```python
median
```

```python
plt.figure(figsize=(10, 6))
plt.barh(df["Feature"], df["LIME Importance"])
plt.xlabel("Aggregated LIME Importance (absolute sum)")
plt.title("Global Feature Importance via LIME")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
median_importance = df["LIME Importance"].median()
# Select features with importance >= mean
selected_features = df[df["LIME Importance"] >= median_importance]["Feature"].tolist()
print(f"Selected {len(selected_features)} features with importance >= median ({median_importance:.4f}):")
print(selected_features)
```
